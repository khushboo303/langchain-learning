```python
# hugging face embedding models
from langchain_huggingface import HuggingFaceEmbeddings

# need to install the model as well, dimensions are default 384
embeddings = HuggingFaceEmbeddings(
    model='sentence-transformers/all-MiniLM-L6-v2')
```

```python
query = "How does Python handle memory?"

documents = [
    "Python uses automatic memory management and garbage collection.",
    "React uses a virtual DOM to efficiently update the UI.",
    "Redis is an in-memory key-value database.",
    "Python lists are mutable ordered collections."
]
```

```python
from sklearn.metrics.pairwise import cosine_similarity

query_vec = embeddings.embed_query(query)
doc_vec = embeddings.embed_documents(documents)

print(query_vec)
print(doc_vec)

# can also directly give two vectors
# for vec in doc_vec:
#     score = cosine_similarity([query_vec], [vec])
#     print(score)

score = cosine_similarity([query_vec], doc_vec)[0]
```

```python
index, value = sorted(list(enumerate(score)), key = lambda x:x[1])[-1]
```

```python
print(index, value)
```

```python
answer = documents[index]
print(answer)
print('similarity score is', value)
```
